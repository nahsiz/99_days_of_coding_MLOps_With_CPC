```python
data = data.copy()

for column in data.columns:
    if data[column].isnull().sum() > 0:
        data[column] = data[column].fillna(data[column].median())

print("Missing values handled.")
```

```python
print("Duplicate rows:", data.duplicated().sum())
```

```python
data = data.drop_duplicates()

print("Updated shape:", data.shape)
```

```python
data.hist(figsize=(15, 10), bins=20)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data['alcohol'], kde=True)

plt.title('Alcohol Distribution')
plt.xlabel('Alcohol')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(5, 3))

sns.boxplot(x=data['alcohol'])

plt.title('Alcohol Boxplot')
plt.show()
```

```python
plt.figure(figsize=(5, 3))

sns.countplot(x='target', data=data)

plt.title('Wine Class Distribution')
plt.xlabel('Target Class')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    data.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=data,
    x='alcohol',
    y='color_intensity',
    hue='target'
)

plt.title('Alcohol vs Color Intensity')
plt.show()
```

```python
data.groupby('target').mean(numeric_only=True)
```
