Cell 1: Import Libraries



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Cell 2: Load Dataset

```python
data = pd.read_csv('/content/day_008_diabetes.csv')

print("Dataset Loaded Successfully!")
data.head()
```

Cell 3: Select Two Columns ( bmi & target )

```python
feature_col = 'bmi'
target_col = 'target'

selected_data = data[[feature_col, target_col]]

selected_data.head()
```

Cell 4: Calculate Correlation

```python
correlation = selected_data[feature_col].corr(
    selected_data[target_col]
)

print("Correlation between", feature_col, "and", target_col)
print("Correlation Coefficient:", correlation)
```

Cell 5: Correlation Heatmap

```python
plt.figure(figsize=(5, 3))

sns.heatmap(
    selected_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap')
plt.show()
```

Cell 6: Scatter Plot with Regression Line ( it shows a relationship and a linear trend.)

```python
plt.figure(figsize=(7, 4))

sns.regplot(
    data=selected_data,
    x=feature_col,
    y=target_col,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{feature_col} vs {target_col}')
plt.xlabel(feature_col)
plt.ylabel(target_col)

plt.show()
```

Cell 7: Histogram of Feature Column

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=data,
    x=feature_col,
    kde=True
)

plt.title(f'Distribution of {feature_col}')
plt.xlabel(feature_col)
plt.ylabel('Frequency')

plt.show()
```

Cell 8: Histogram of Target Column

```python
plt.figure(figsize=(6, 4))

sns.histplot(
    data=data,
    x=target_col,
    kde=True
)

plt.title(f'Distribution of {target_col}')
plt.xlabel(target_col)
plt.ylabel('Frequency')

plt.show()
```

Cell 9: Joint Plot ( scatter distribution and individual distributions )

```python
sns.jointplot(
    data=selected_data,
    x=feature_col,
    y=target_col,
    kind='reg',
    height=7
)

plt.show()
```

Cell 10: Correlation Interpretation

```python
if correlation > 0:
    print("Positive Correlation")
elif correlation < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

print("Correlation Value:", round(correlation, 4))
```
