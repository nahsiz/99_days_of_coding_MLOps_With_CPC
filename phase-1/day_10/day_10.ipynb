```python
data = pd.read_csv('/content/day_010_house_price.csv')

print("Dataset Loaded Successfully!")
```

Part 2 — Selecting Three Columns

8. Select Three Target Columns

Text Box Title: Selecting Three Columns for Correlation Analysis

Text Box Description: Select two feature columns (area_sqft, distance_km) and one target column (price) to investigate the relationship between each feature and the target.

```python
feature_col_1 = 'area_sqft'
feature_col_2 = 'distance_km'
target_col = 'price'

selected_data = data[[feature_col_1, feature_col_2, target_col]]

selected_data.head()
```

9. Check Selected Columns

Text Box Title: Inspecting the Selected Columns

Text Box Description: Display the selected columns and check their data types and statistical information before calculating correlation.

```python
print(selected_data.info())

print("\nStatistical Summary:")
display(selected_data.describe())
```

10. Calculate Pearson Correlation

Text Box Title: Calculating Correlation Between Three Columns

Text Box Description: Calculate the Pearson correlation coefficient twice — once for area_sqft vs price, and once for distance_km vs price.

```python
correlation_1 = selected_data[feature_col_1].corr(
    selected_data[target_col]
)

correlation_2 = selected_data[feature_col_2].corr(
    selected_data[target_col]
)

print("Correlation between", feature_col_1, "and", target_col)
print("Pearson Correlation Coefficient:", round(correlation_1, 4))

print()

print("Correlation between", feature_col_2, "and", target_col)
print("Pearson Correlation Coefficient:", round(correlation_2, 4))
```

Part 3 — Correlation Visualization

11. Correlation Heatmap

Text Box Title: Visualizing Correlation Using a Heatmap

Text Box Description: Create a heatmap to display the correlation coefficients between the three selected columns.

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    selected_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap')
plt.show()
```

12. Scatter Plot with Regression Line — First Feature

Text Box Title: Visualizing the Relationship for the First Feature

Text Box Description: Create a scatter plot with a regression line to observe the relationship between area_sqft and price.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=feature_col_1,
    y=target_col,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{feature_col_1} vs {target_col}')
plt.xlabel(feature_col_1)
plt.ylabel(target_col)
plt.show()
```

13. Scatter Plot with Regression Line — Second Feature

Text Box Title: Visualizing the Relationship for the Second Feature

Text Box Description: Create a scatter plot with a regression line to observe the relationship between distance_km and price. The y-axis stays the same (price) so the two plots can be compared.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=feature_col_2,
    y=target_col,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{feature_col_2} vs {target_col}')
plt.xlabel(feature_col_2)
plt.ylabel(target_col)
plt.show()
```

14. Histogram of the First Column

Text Box Title: Distribution of Area

Text Box Description: Use a histogram with a KDE curve to understand the distribution of area_sqft.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=feature_col_1,
    kde=True
)

plt.title(f'Distribution of {feature_col_1}')
plt.xlabel(feature_col_1)
plt.ylabel('Frequency')
plt.show()
```

15. Histogram of the Second Column

Text Box Title: Distribution of Distance

Text Box Description: Use a histogram with a KDE curve to understand the distribution of distance_km.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=feature_col_2,
    kde=True
)

plt.title(f'Distribution of {feature_col_2}')
plt.xlabel(feature_col_2)
plt.ylabel('Frequency')
plt.show()
```

16. Histogram of the Third Column

Text Box Title: Distribution of Price

Text Box Description: Use a histogram with a KDE curve to understand the distribution of price.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=target_col,
    kde=True
)

plt.title(f'Distribution of {target_col}')
plt.xlabel(target_col)
plt.ylabel('Frequency')
plt.show()
```

17. Joint Plot — First Feature

Text Box Title: Combined Visualization of the First Feature and Target

Text Box Description: Use a joint plot to visualize the relationship between area_sqft and price along with their individual distributions.

```python
sns.jointplot(
    data=selected_data,
    x=feature_col_1,
    y=target_col,
    kind='reg',
    height=7
)

plt.show()
```

18. Joint Plot — Second Feature

Text Box Title: Combined Visualization of the Second Feature and Target

Text Box Description: Use a joint plot to visualize the relationship between distance_km and price along with their individual distributions.

```python
sns.jointplot(
    data=selected_data,
    x=feature_col_2,
    y=target_col,
    kind='reg',
    height=7
)

plt.show()
```

**Part 4 — Correlation Interpretation**

19. Interpret the First Correlation Result

Text Box Title: Interpreting the First Correlation Coefficient

Text Box Description: Identify whether the relationship between area_sqft and price is positive, negative, or has no linear correlation.

```python
if correlation_1 > 0:
    print("Positive Correlation")
elif correlation_1 < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

print("Correlation Value:", round(correlation_1, 4))
```

20. Interpret the Second Correlation Result

Text Box Title: Interpreting the Second Correlation Coefficient

Text Box Description: Identify whether the relationship between distance_km and price is positive, negative, or has no linear correlation.

```python
if correlation_2 > 0:
    print("Positive Correlation")
elif correlation_2 < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

print("Correlation Value:", round(correlation_2, 4))
```

21. Display Final Correlation Result

Text Box Title: Final Correlation Analysis Result

Text Box Description: Display the selected column names and their calculated Pearson correlation coefficients.

```python
print("Final Correlation Analysis")
print("--------------------------")

print("First Column:", feature_col_1)
print("Target Column:", target_col)
print("Pearson Correlation:", round(correlation_1, 4))

print()

print("Second Column:", feature_col_2)
print("Target Column:", target_col)
print("Pearson Correlation:", round(correlation_2, 4))
```
