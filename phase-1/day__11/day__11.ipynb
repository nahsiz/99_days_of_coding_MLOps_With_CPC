Part 2 — Selecting Three Columns

8. Select Three Columns (Shared X-Axis)

Text Box Title: Selecting Three Columns for Correlation Analysis

Text Box Description: Select one shared x-axis column (Kms_Driven) and two y-axis columns (Selling_Price, Present_Price). Kms_Driven will stay on the x-axis in both scatter plots so the two relationships can be compared directly against the same horizontal axis.

```python
x_col = 'Kms_Driven'
y_col_1 = 'Selling_Price'
y_col_2 = 'Present_Price'

selected_data = data[[x_col, y_col_1, y_col_2]]

selected_data.head()
```

9. Check Selected Columns

Text Box Title: Inspecting the Selected Columns

Text Box Description: Display the selected columns and check their data types and statistical information before calculating correlation.

```python
print(selected_data.info())

print("\nStatistical Summary:")
display(selected_data.describe())
```

10. Calculate Pearson Correlation

Text Box Title: Calculating Correlation Between Three Columns

Text Box Description: Calculate the Pearson correlation coefficient twice — once for Kms_Driven vs Selling_Price, and once for Kms_Driven vs Present_Price. Kms_Driven remains the shared column in both calculations, matching the shared x-axis requirement.

```python
correlation_1 = selected_data[x_col].corr(
    selected_data[y_col_1]
)

correlation_2 = selected_data[x_col].corr(
    selected_data[y_col_2]
)

print("Correlation between", x_col, "and", y_col_1)
print("Pearson Correlation Coefficient:", round(correlation_1, 4))

print()

print("Correlation between", x_col, "and", y_col_2)
print("Pearson Correlation Coefficient:", round(correlation_2, 4))
```

Part 3 — Correlation Visualization

11. Correlation Heatmap

Text Box Title: Visualizing Correlation Using a Heatmap

Text Box Description: Create a heatmap to display the correlation coefficients between the three selected columns.

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    selected_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap')
plt.show()
```

12. Scatter Plot — Shared X-Axis with First Y

Text Box Title: Visualizing Kms_Driven vs Selling_Price

Text Box Description: Create a scatter plot with a regression line where Kms_Driven is on the x-axis and Selling_Price is on the y-axis. Kms_Driven stays on the x-axis here because it will be the shared axis for both plots.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=x_col,
    y=y_col_1,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{x_col} vs {y_col_1}')
plt.xlabel(x_col)
plt.ylabel(y_col_1)
plt.show()
```

13. Scatter Plot — Shared X-Axis with Second Y

Text Box Title: Visualizing Kms_Driven vs Present_Price

Text Box Description: Create a scatter plot with a regression line where Kms_Driven is again on the x-axis, but Present_Price is now on the y-axis. The x-axis stays identical to the previous plot so both relationships can be compared on the same horizontal axis.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=x_col,
    y=y_col_2,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{x_col} vs {y_col_2}')
plt.xlabel(x_col)
plt.ylabel(y_col_2)
plt.show()
```

14. Histogram of the First Column (Shared X)

Text Box Title: Distribution of Kms_Driven

Text Box Description: Use a histogram with a KDE curve to understand the distribution of the Kms_Driven column.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=x_col,
    kde=True
)

plt.title(f'Distribution of {x_col}')
plt.xlabel(x_col)
plt.ylabel('Frequency')
plt.show()
```

15. Histogram of the Second Column

Text Box Title: Distribution of Selling_Price

Text Box Description: Use a histogram with a KDE curve to understand the distribution of the Selling_Price column.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=y_col_1,
    kde=True
)

plt.title(f'Distribution of {y_col_1}')
plt.xlabel(y_col_1)
plt.ylabel('Frequency')
plt.show()
```

16. Histogram of the Third Column

Text Box Title: Distribution of Present_Price

Text Box Description: Use a histogram with a KDE curve to understand the distribution of the Present_Price column.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=y_col_2,
    kde=True
)

plt.title(f'Distribution of {y_col_2}')
plt.xlabel(y_col_2)
plt.ylabel('Frequency')
plt.show()
```

17. Joint Plot — Shared X with First Y

Text Box Title: Combined Visualization of Kms_Driven and Selling_Price

Text Box Description: Use a joint plot to visualize the relationship between Kms_Driven and Selling_Price along with their individual distributions.

```python
sns.jointplot(
    data=selected_data,
    x=x_col,
    y=y_col_1,
    kind='reg',
    height=7
)

plt.show()
```

18. Joint Plot — Shared X with Second Y

Text Box Title: Combined Visualization of Kms_Driven and Present_Price

Text Box Description: Use a joint plot to visualize the relationship between Kms_Driven and Present_Price along with their individual distributions. The x-axis stays Kms_Driven again so both joint plots share the same horizontal axis.

```python
sns.jointplot(
    data=selected_data,
    x=x_col,
    y=y_col_2,
    kind='reg',
    height=7
)

plt.show()
```

Part 4 — Correlation Interpretation

19. Interpret the First Correlation Result

Text Box Title: Interpreting the First Correlation Coefficient

Text Box Description: Identify whether the relationship between Kms_Driven and Selling_Price is positive, negative, or has no linear correlation.

```python
if correlation_1 > 0:
    print("Positive Correlation")
elif correlation_1 < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

print("Correlation Value:", round(correlation_1, 4))
```

20. Interpret the Second Correlation Result

Text Box Title: Interpreting the Second Correlation Coefficient

Text Box Description: Identify whether the relationship between Kms_Driven and Present_Price is positive, negative, or has no linear correlation.

```python
if correlation_2 > 0:
    print("Positive Correlation")
elif correlation_2 < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

print("Correlation Value:", round(correlation_2, 4))
```

Text Box Title: Final Correlation Analysis Result

Text Box Description: Display the shared x-axis column name and both y-axis column names along with their calculated Pearson correlation coefficients.

```python
print("Final Correlation Analysis")
print("--------------------------")

print("Shared X-Axis Column:", x_col)
print("First Y-Axis Column :", y_col_1)
print("Pearson Correlation :", round(correlation_1, 4))

print()

print("Shared X-Axis Column:", x_col)
print("Second Y-Axis Column:", y_col_2)
print("Pearson Correlation :", round(correlation_2, 4))
```
