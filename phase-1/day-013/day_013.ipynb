Part 2 — Selecting Four Columns

8. Select Four Columns (Shared X-Axis)

Text Box Title: Selecting Four Columns for Correlation Analysis

Text Box Description: Select one shared x-axis column (age) and three y-axis columns (website_visits, purchase_frequency, total_spend). The age column will stay on the x-axis in all scatter plots so the three relationships can be compared against the same horizontal axis.

```python
x_col = 'age'
y_col_1 = 'website_visits'
y_col_2 = 'purchase_frequency'
y_col_3 = 'total_spend'

selected_data = data[[x_col, y_col_1, y_col_2, y_col_3]]

selected_data.head()
```

9. Check Selected Columns

Text Box Title: Inspecting the Selected Columns

Text Box Description: Display the selected columns and check their data types and statistical information before calculating correlation.

```python
print(selected_data.info())

print("\nStatistical Summary:")
display(selected_data.describe())
```

10. Calculate Pearson Correlation

Text Box Title: Calculating Correlation Between Four Columns

Text Box Description: Calculate the Pearson correlation coefficient three times — age vs website_visits, age vs purchase_frequency, and age vs total_spend. The age column remains the shared x-axis column in all three calculations.

```python
correlation_1 = selected_data[x_col].corr(
    selected_data[y_col_1]
)

correlation_2 = selected_data[x_col].corr(
    selected_data[y_col_2]
)

correlation_3 = selected_data[x_col].corr(
    selected_data[y_col_3]
)

print("Correlation between", x_col, "and", y_col_1)
print("Pearson Correlation Coefficient:", round(correlation_1, 4))

print()

print("Correlation between", x_col, "and", y_col_2)
print("Pearson Correlation Coefficient:", round(correlation_2, 4))

print()

print("Correlation between", x_col, "and", y_col_3)
print("Pearson Correlation Coefficient:", round(correlation_3, 4))
```

Part 3 — Correlation Visualization

11. Correlation Heatmap (4×4)

Text Box Title: Visualizing Correlation Using a Heatmap

Text Box Description: Create a 4×4 heatmap to display the correlation coefficients between all four selected columns.

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    selected_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap')
plt.show()
```

12. Scatter Plot — Shared X with First Y

Text Box Title: Visualizing Age vs Website Visits

Text Box Description: Create a scatter plot with a regression line where age is on the x-axis and website_visits is on the y-axis. The age column stays on the x-axis here because it will be the shared axis for all three plots.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=x_col,
    y=y_col_1,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{x_col} vs {y_col_1}')
plt.xlabel(x_col)
plt.ylabel(y_col_1)
plt.show()
```

13. Scatter Plot — Shared X with Second Y

Text Box Title: Visualizing Age vs Purchase Frequency

Text Box Description: Create a scatter plot with a regression line where age is again on the x-axis, but purchase_frequency is now on the y-axis. The x-axis stays identical to the previous plot so the relationships can be compared on the same horizontal axis.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=x_col,
    y=y_col_2,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{x_col} vs {y_col_2}')
plt.xlabel(x_col)
plt.ylabel(y_col_2)
plt.show()
```

14. Scatter Plot — Shared X with Third Y

Text Box Title: Visualizing Age vs Total Spend

Text Box Description: Create a scatter plot with a regression line where age is again on the x-axis, but total_spend is now on the y-axis. The x-axis stays identical so all three plots share the same horizontal axis.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=x_col,
    y=y_col_3,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{x_col} vs {y_col_3}')
plt.xlabel(x_col)
plt.ylabel(y_col_3)
plt.show()
```
