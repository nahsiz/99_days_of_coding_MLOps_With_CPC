1. Import Required Libraries

2. Load the Dataset
Text Box Title: Loading the Student Performance Dataset

Text Box Description: Load the CSV file into a Pandas DataFrame for further analysis.

```python
data = pd.read_csv('/content/day_009_student_performance.csv')

print("Dataset Loaded Successfully!")
```

**Part 2 — Selecting Two Columns**

8. Select Two Target Columns
Text Box Title: Selecting Two Columns for Correlation Analysis

Text Box Description: Select study_hours and exam_score to investigate the relationship between study time and exam performance.

```python
feature_col = 'study_hours'
target_col = 'exam_score'

selected_data = data[[feature_col, target_col]]

selected_data.head()
```

9. Check Selected Columns
Text Box Title: Inspecting the Selected Columns

Text Box Description: Display the selected columns and check their data types and statistical information before calculating correlation.

```python
print(selected_data.info())

print("\nStatistical Summary:")
display(selected_data.describe())
```

10. Calculate Pearson Correlation
Text Box Title: Calculating Correlation Between Two Columns

Text Box Description: Calculate the Pearson correlation coefficient to measure the strength and direction of the linear relationship between study_hours and exam_score.

```python
correlation = selected_data[feature_col].corr(
    selected_data[target_col]
)

print("Correlation between", feature_col, "and", target_col)
print("Pearson Correlation Coefficient:", round(correlation, 4))
```

**Part 3 — Correlation Visualization**

11. Correlation Heatmap
Text Box Title: Visualizing Correlation Using a Heatmap

Text Box Description: Create a heatmap to display the correlation coefficient between the two selected columns.

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    selected_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap')
plt.show()
```

12. Scatter Plot with Regression Line
Text Box Title: Visualizing the Relationship Using a Scatter Plot

Text Box Description: Create a scatter plot with a regression line to observe the relationship and linear trend between study hours and exam scores.

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=selected_data,
    x=feature_col,
    y=target_col,
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'red'}
)

plt.title(f'{feature_col} vs {target_col}')
plt.xlabel(feature_col)
plt.ylabel(target_col)

plt.show()
```

13. Histogram of the First Column

Text Box Title: Distribution of Study Hours

Text Box Description: Use a histogram with a KDE curve to understand the distribution of study hours.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=feature_col,
    kde=True
)

plt.title(f'Distribution of {feature_col}')
plt.xlabel(feature_col)
plt.ylabel('Frequency')

plt.show()
```

14. Histogram of the Second Column

Text Box Title: Distribution of Exam Scores

Text Box Description: Use a histogram with a KDE curve to understand the distribution of exam scores.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x=target_col,
    kde=True
)

plt.title(f'Distribution of {target_col}')
plt.xlabel(target_col)
plt.ylabel('Frequency')

plt.show()
```

15. Joint Plot

Text Box Title: Combined Visualization of Two Columns

Text Box Description: Use a joint plot to visualize the relationship between the two selected columns along with their individual distributions.

```python
sns.jointplot(
    data=selected_data,
    x=feature_col,
    y=target_col,
    kind='reg',
    height=7
)

plt.show()
```

Part 4 — Correlation Interpretation

16. Interpret the Correlation Result

Text Box Title: Interpreting the Correlation Coefficient

Text Box Description: Identify whether the relationship between the selected columns is positive, negative, or has no linear correlation based on the Pearson correlation coefficient.

```python
if correlation > 0:
    print("Positive Correlation")
elif correlation < 0:
    print("Negative Correlation")
else:
    print("No Linear Correlation")

print("Correlation Value:", round(correlation, 4))
```

17. Display Final Correlation Result

Text Box Title: Final Correlation Analysis Result

Text Box Description: Display the selected column names and their calculated Pearson correlation coefficient.



```python
print("Final Correlation Analysis")
print("--------------------------")
print("First Column:", feature_col)
print("Second Column:", target_col)
print("Pearson Correlation:", round(correlation, 4))
```
